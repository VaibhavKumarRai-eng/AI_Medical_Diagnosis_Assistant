# Exploratory Data Analysis (EDA) - AI Medical Diagnosis Assistant

This notebook conducts a comprehensive Exploratory Data Analysis on the clinical medical symptom dataset to inspect schema types, class distributions, symptom text length statistics, missing values, duplicates, and n-gram word frequencies.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter

# Set modern visual styling
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette('muted')
```

## 1. Data Loading & Inspection

```python
dataset_path = os.path.join('..', 'dataset', 'final_symptoms_to_disease.csv')
df = pd.read_csv(dataset_path)

print(f"Dataset Shape: {df.shape}")
print("\nData Types:")
print(df.dtypes)
print("\nFirst 5 Rows:")
df.head()
```

## 2. Missing Values & Duplicate Rows

```python
print("Missing Values:")
print(df.isnull().sum())

print(f"\nDuplicate Rows: {df.duplicated().sum()}")
```

## 3. Class Distribution & Top Diseases

```python
disease_counts = df['diseases'].value_counts()
print(f"Total Unique Diseases: {len(disease_counts)}")

plt.figure(figsize=(12, 6))
sns.barplot(x=disease_counts.head(20).values, y=disease_counts.head(20).index, palette='Blues_r')
plt.title('Top 20 Most Frequent Disease Categories', fontsize=14, fontweight='bold')
plt.xlabel('Sample Count')
plt.ylabel('Disease Name')
plt.tight_layout()
plt.show()
```

## 4. Symptom Text Length Distribution

```python
df['char_length'] = df['symptom_text'].astype(str).apply(len)
df['word_count'] = df['symptom_text'].astype(str).apply(lambda x: len(x.split()))

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.histplot(df['char_length'], bins=50, kde=True, ax=axes[0], color='#2b5c8f')
axes[0].set_title('Symptom Character Length Distribution', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Character Count')

sns.histplot(df['word_count'], bins=50, kde=True, ax=axes[1], color='#2ecc71')
axes[1].set_title('Symptom Word Count Distribution', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Word Count')

plt.tight_layout()
plt.show()
```

## 5. Most Frequent Symptom Words Analysis

```python
all_words = ' '.join(df['symptom_text'].dropna().astype(str)).lower().split()
word_freq = Counter(all_words)
common_words = pd.DataFrame(word_freq.most_common(20), columns=['Word', 'Count'])

plt.figure(figsize=(12, 6))
sns.barplot(data=common_words, x='Count', y='Word', palette='viridis')
plt.title('Top 20 Most Frequent Words in Symptoms', fontsize=14, fontweight='bold')
plt.xlabel('Frequency Count')
plt.tight_layout()
plt.show()
```
